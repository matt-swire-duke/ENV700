````python
# Part A

import math

pi = (math.pi)

radius = 20
def circle_area(radius):
     area = pi * radius ** 2 
     return area

area = circle_area(radius)

rounded_area = round(area,2)

print('Rounded area of circle:',rounded_area)
````

````python
#Part B Function with 2 inputs
length = 191
width = 1051

def recentagle_area(length,width):
    area = length * width 
    return area 

area = recentagle_area(length, width)

print(area)
````

````python
#Part C 
number_list = [41,51,25,867,5211,236,11,-1]

def min_max_values(number_list):
    min_list_value = min(number_list)
    max_list_value = max(number_list)
    return min_list_value, max_list_value

min_list_value, max_list_value = min_max_values(number_list)

print('Min value:',min_list_value)
print('Max value:',max_list_value)
````
