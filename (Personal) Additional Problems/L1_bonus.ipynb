## Problem 4: Track Unique Species (Sets)

You observe the following species at your monitoring site over several days:
deer, squirrel, deer, rabbit, squirrel, deer, fox, rabbit

a) Create a set called `observed_species` with these observations (notice duplicates should be removed)

b) Print the set and its length

c) How many unique species did you observe?

````python
# Problem 4: Track Unique Species
# Write your code here

observed_species = ['deer', 'squirrel', 'deer', 'rabbit', 'squirrel', 'deer', 'fox', 'rabbit']
clean_observed_species = []
for species in observed_species:
    if species not in clean_observed_species:
        clean_observed_species.append(species)


print(clean_observed_species)
print(len(clean_observed_species))
````

## Problem 5: Water Quality Classification (Conditionals)

## Problem 6: Analyze Rainfall Data (Loops & Conditionals)

You have 10 days of rainfall measurements (in mm):
[0.0, 2.3, 0.0, 5.1, 0.0, 0.0, 3.8, 1.2, 0.0, 4.5]

a) Loop through and print only the days with rainfall > 0 mm

b) Count how many days had no rainfall (rainfall = 0)

c) Count how many days had significant rainfall (> 2.0 mm)

**Hint:** Use a for loop and if statements. You may want to track a counter variable.

````python
# Problem 6: Analyze Rainfall Data
# Write your code here

rainfall = [0.0, 2.3, 0.0, 5.1, 0.0, 0.0, 3.8, 1.2, 0.0, 4.5]
rainfall_days = 0
significant_rainfall_days = 0

for day in rainfall:
    if day > 2:
        significant_rainfall_days += 1
        rainfall_days += 1
    elif day > 0:
        rainfall_days += 1

print(rainfall_days)
print(significant_rainfall_days)

````

## Problem 7: Use range() and break (Loops with Control)

## Problem 9: Challenge - Weekly Summary Report (Combining Everything)

Given this week's monitoring data:

```python
daily_temps_c = [18.5, 19.2, 21.4, 20.8, 19.3, 17.6, 16.9]
daily_rainfall_mm = [0.0, 2.3, 0.0, 5.1, 0.0, 0.0, 3.8]
station_info = {"name": "Duke Station", "elevation_m": 120, "state": "NC"}

```

Write a program that:

1. Loops through temperatures and counts days above 20°C
2. Loops through rainfall and calculates total rainfall for the week
3. Calculates average temperature for the week
4. Prints a formatted report:

```
=== WEEKLY MONITORING REPORT ===
Station: [name] - [state] at [elevation]m
Days above 20°C: [count]
Total rainfall: [total]mm
Average temperature: [avg]°C
Unique species observed: [count]
Species: [list species]
```

````python
# Problem 9: Challenge - Weekly Summary Report
# Given data:

daily_temps_c = [18.5, 19.2, 21.4, 20.8, 19.3, 17.6, 16.9]
daily_rainfall_mm = [0.0, 2.3, 0.0, 5.1, 0.0, 0.0, 3.8]
station_info = {
    "name": "Duke Station", 
    "elevation_m": 120, 
    "state": "NC"
    }
observed_species = {"deer", "squirrel", "rabbit", "fox"}


days_above_20 = 0 
for day in daily_temps_c:
    if day > 20:
        days_above_20 += 1 



total_rainfall_mm = sum(daily_rainfall_mm)
average_temp = round(sum(daily_temps_c) / len(daily_temps_c),2)

print(f"""=== WEEKLY MONITORING REPORT ===
Station: {station_info["name"]} - {station_info["state"]} at {station_info["elevation_m"]}m
Days above 20°C: {days_above_20}
Total rainfall: {total_rainfall_mm}mm
Average temperature: {average_temp}°C
Unique species observed: {len(observed_species)}
Species: {observed_species}""")
````
